```python
import sys
import os

parent_dir = os.path.abspath(os.path.join(os.getcwd(), ".."))
if parent_dir not in sys.path:
    sys.path.append(parent_dir)

from pcdqv.standard_regularization import RandomizedHadamard
```

# Test 1

```python
w = torch.arange(0, 16).reshape(4, 4).to(torch.float64)
w = w * 1.1
print(f"{w=}")

regularizer = RandomizedHadamard(
    p = w.shape[1],
)
```

```python
w_forward = regularizer.forward(w)
w_forward
```

```python
w_revesed = regularizer.reverse(w_forward)
w_revesed
```

```python
assert torch.allclose(
    w,
    regularizer.reverse(w_forward),
)
```
